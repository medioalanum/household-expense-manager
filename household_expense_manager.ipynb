Gestore delle spese domestiche

Questo programma serve a tenere traccia delle spese di casa. Puoi aggiungere una spesa, vedere quanto hai speso ogni mese e consultare le dieci spese più alte. Per ogni spesa vengono salvati la data, la descrizione e l’importo.

Per iniziare, esegui la prima cella di codice e poi la seconda. Scegli un’opzione dal menu e segui le istruzioni. Inserisci 0 per uscire. Per usare di nuovo il programma, basta eseguire la seconda cella.

Inserisci le date nel formato GG/MM/AAAA, per esempio 18/05/2024. Per gli importi puoi scrivere 12.50 oppure 12,50. Usa sempre la stessa valuta e non inserire separatori delle migliaia.

Expenses are saved in expenses.csv and remain available when you run the program again. Colab files are temporary. Download expenses.csv from the Files panel before ending the session if you want to keep the data.

```python
import csv
import os
from datetime import datetime


FILE_NAME = "expenses.csv"


# Create the CSV file only when it does not exist.
def prepare_file():
    if not os.path.exists(FILE_NAME):
        with open(FILE_NAME, "w", newline="", encoding="utf-8") as file:
            writer = csv.writer(file)
            writer.writerow(["date", "description", "amount"])


# Read the saved expenses and convert the amounts to numbers.
def read_expenses():
    prepare_file()
    expenses = []
    with open(FILE_NAME, "r", newline="", encoding="utf-8") as file:
        reader = csv.DictReader(file)
        for row in reader:
            row["amount"] = float(row["amount"])
            expenses.append(row)
    return expenses


# Check the user's entries before saving a new expense.
def add_expense():
    while True:
        date = input("Data (GG/MM/AAAA): ").strip()
        try:
            date = datetime.strptime(date, "%d/%m/%Y").strftime("%d/%m/%Y")
            break
        except ValueError:
            print("Inserisci una data valida nel formato GG/MM/AAAA.")

    description = input("Descrizione: ").strip()
    while not description:
        print("La descrizione non può essere vuota.")
        description = input("Descrizione: ").strip()

    while True:
        try:
            amount = float(input("Importo: ").strip().replace(",", "."))
            if not 0 < amount < float("inf"):
                print("Inserisci un importo positivo e finito.")
                continue
            amount = round(amount, 2)
            if amount == 0:
                print("L'importo deve essere almeno 0,01.")
                continue
            break
        except ValueError:
            print("Inserisci un numero valido, per esempio 12,50.")

    prepare_file()
    with open(FILE_NAME, "a", newline="", encoding="utf-8") as file:
        writer = csv.writer(file)
        writer.writerow([date, description, f"{amount:.2f}"])
    print("Spesa salvata.")


# Add together the expenses for each year and month.
def show_monthly_report():
    expenses = read_expenses()
    if not expenses:
        print("Non ci sono ancora spese registrate.")
        return

    totals = {}
    for expense in expenses:
        date = datetime.strptime(expense["date"], "%d/%m/%Y")
        month = date.strftime("%Y-%m")
        if month not in totals:
            totals[month] = 0
        totals[month] += expense["amount"]

    print("Riepilogo mensile")
    for month in sorted(totals):
        print(f"{month}: {totals[month]:.2f}")


# Use the numeric amount to sort expenses from highest to lowest.
def get_amount(expense):
    return expense["amount"]


def show_top_expenses():
    expenses = read_expenses()
    if not expenses:
        print("Non ci sono ancora spese registrate.")
        return

    expenses = sorted(expenses, key=get_amount, reverse=True)
    print("Le 10 spese più alte")
    for expense in expenses[:10]:
        print(f"{expense['date']} | {expense['description']} | {expense['amount']:.2f}")


# Keep showing the menu until the user chooses to exit.
def main():
    prepare_file()
    while True:
        print("\nGestore delle spese domestiche")
        print("1. Aggiungi una spesa")
        print("2. Riepilogo mensile")
        print("3. Le 10 spese più alte")
        print("0. Esci")
        choice = input("Scegli un'opzione: ").strip()

        if choice == "1":
            add_expense()
        elif choice == "2":
            show_monthly_report()
        elif choice == "3":
            show_top_expenses()
        elif choice == "0":
            print("Arrivederci.")
            break
        else:
            print("Opzione non valida. Riprova.")
```

```python
# Start the menu. Choose 0 to stop the program.
main()
```
