Household Expense Manager

This program helps track household expenses. You can add an expense, view monthly spending, and review the ten highest expenses. Each expense stores a date, description, and amount.

To start, run the first code cell and then the second one. Choose an option from the menu and follow the instructions. Enter 0 to exit. To use the program again, run the second cell.

Enter dates in DD/MM/YYYY format, for example 18/05/2024. For amounts, enter 12.50 or 12,50. Use the same currency and do not enter thousands separators.

Expenses are saved in expenses.csv and remain available when you run the program again. Colab files are temporary. Download expenses.csv from the Files panel before ending the session if you want to keep the data.

```python
import csv
import os
from datetime import datetime


FILE_NAME = "expenses.csv"


# Create the CSV file only when it does not exist.
def prepare_file():
    if not os.path.exists(FILE_NAME):
        with open(FILE_NAME, "w", newline="", encoding="utf-8") as file:
            writer = csv.writer(file)
            writer.writerow(["date", "description", "amount"])


# Read the saved expenses and convert the amounts to numbers.
def read_expenses():
    prepare_file()
    expenses = []
    with open(FILE_NAME, "r", newline="", encoding="utf-8") as file:
        reader = csv.DictReader(file)
        for row in reader:
            row["amount"] = float(row["amount"])
            expenses.append(row)
    return expenses


# Check the user's entries before saving a new expense.
def add_expense():
    while True:
        date = input("Date (DD/MM/YYYY): ").strip()
        try:
            date = datetime.strptime(date, "%d/%m/%Y").strftime("%d/%m/%Y")
            break
        except ValueError:
            print("Enter a valid date in DD/MM/YYYY format.")

    description = input("Description: ").strip()
    while not description:
        print("Description cannot be empty.")
        description = input("Description: ").strip()

    while True:
        try:
            amount = float(input("Amount: ").strip().replace(",", "."))
            if not 0 < amount < float("inf"):
                print("Enter a positive, finite amount.")
                continue
            amount = round(amount, 2)
            if amount == 0:
                print("The amount must be at least 0.01.")
                continue
            break
        except ValueError:
            print("Enter a valid number, for example 12.50.")

    prepare_file()
    with open(FILE_NAME, "a", newline="", encoding="utf-8") as file:
        writer = csv.writer(file)
        writer.writerow([date, description, f"{amount:.2f}"])
    print("Expense saved.")


# Add together the expenses for each year and month.
def show_monthly_report():
    expenses = read_expenses()
    if not expenses:
        print("No expenses have been recorded yet.")
        return

    totals = {}
    for expense in expenses:
        date = datetime.strptime(expense["date"], "%d/%m/%Y")
        month = date.strftime("%Y-%m")
        if month not in totals:
            totals[month] = 0
        totals[month] += expense["amount"]

    print("Monthly report")
    for month in sorted(totals):
        print(f"{month}: {totals[month]:.2f}")


# Use the numeric amount to sort expenses from highest to lowest.
def get_amount(expense):
    return expense["amount"]


def show_top_expenses():
    expenses = read_expenses()
    if not expenses:
        print("No expenses have been recorded yet.")
        return

    expenses = sorted(expenses, key=get_amount, reverse=True)
    print("Top 10 expenses")
    for expense in expenses[:10]:
        print(f"{expense['date']} | {expense['description']} | {expense['amount']:.2f}")


# Keep showing the menu until the user chooses to exit.
def main():
    prepare_file()
    while True:
        print("\nHousehold Expense Manager")
        print("1. Add an expense")
        print("2. Monthly report")
        print("3. Top 10 expenses")
        print("0. Exit")
        choice = input("Choose an option: ").strip()

        if choice == "1":
            add_expense()
        elif choice == "2":
            show_monthly_report()
        elif choice == "3":
            show_top_expenses()
        elif choice == "0":
            print("Goodbye.")
            break
        else:
            print("Invalid option. Try again.")
```

```python
# Start the menu. Choose 0 to stop the program.
main()
```
